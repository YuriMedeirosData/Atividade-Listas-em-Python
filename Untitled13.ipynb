```python
# como funciona o while?
# while condição:

# exemplo 1 - contar até o usuaário pedir

limite = int(input("Informe um número limite: "))
inicio = 0

while inicio <= limite:
  print(inicio)
  inicio = inicio + 1
```

```python
# controle com while

senha_correta = 'abc123'
tentativas = 0

while tentativas < 3:
  senha = input("Insira a sua senha: ")
  if senha_correta == senha:
    print("Bem-vindo ao sistema!")
    break
  else:
    print("Senha incorreta! \nTente de novo")
    tentativas = tentativas + 1
    print(f"Número de tentativas: {tentativas}")
```

```python
# while true

while True:
  comando = input("Para sair, digite a palavra SAIR: ")
  if comando.lower() == "sair":
    print("Até mais!")
    break
  else:
    print("Continuando o sistema...")
```

```python
# atividade 1 - Fácil

# crie um sistema que pergunte ao usuário se ele quer
# salvar uma informação numa lista de carros

# a lista --> carros = ["uno", "corolla", "fusca"]

carros = ["uno", "corolla", "fusca"]

escolha = str(input("Deseja adicionar um novo carro? SIM ou NÃO: ")).lower()

while escolha == 'sim':
  novo_carro = str(input("Cadastre o novo carro: "))
  carros.append(novo_carro)

  escolha = str(input("Deseja cadastrar um novo carro? SIM ou NÃO: ")).lower()

print(f"Lista de carros: {carros}")
```

```python
carros = ["uno", "corolla", "fusca"]

print("-=-=-=-= BEM-VINDOS -=-=-=-=")

while True:
  novoCarro = str(input("Informe um novo carro.\nDigite SAIR para finalizar: "))
  if novoCarro == "sair":
    print("Até mais!")
    break

  else:
    carros.append(novoCarro)
    for i in carros:
      print(i)
```
