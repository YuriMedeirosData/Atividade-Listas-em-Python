```python
# PARTE 1 - CRIANDO LISTAS INICIAIS
produtos = ['camisa básica', 'calça-jeans', 'jaqueta corta-vento']
precos = [49.98, 129.90, 219.90]
```

```python
# PARTE 2 - CADASTRANDO NOVOS ITENS

novo_produto = input("Deseja cadastrar um novo produto? ")

if novo_produto.lower() == "sim":
  nova_peca = str(input("Digite o novo produto: "))
  novo_preco = float(input("Preço do novo produto: R$"))

produtos.append(nova_peca)
precos.append(novo_preco)

for i, j in zip(produtos, precos):
  print(f"O produto {i} custa R${j:.2f}")
```

```python
# PARTE 3 - REMOVER O ULTIMO ITEM (POP)

remover = len(produtos)

if remover > 0:
  produtos.pop()
  precos.pop()

  print("O ultimo produto removido")

  for i, j in zip(produtos, precos):
    print(f"O produto {i} custa R${j:.2f}")
else:
  print("Estoque vazio!")
```

```python
# PARTE 4 - BUSCA E REMOÇÃO POR NOME

item = input("Deseja remover algum produto? ")

if item in produtos:
  indice = produtos.index(item)
  produtos.pop(indice)
  precos.pop(indice)

  for i, j in zip(produtos, precos):
    print(f"O produto {i} custa R${j:.2f}")

else:
  print("Produto não encontrado!")





```
